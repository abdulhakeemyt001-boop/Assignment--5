**Q1:** Aik `Student` class banayein jismein `name` aur `roll_number` private variables (`__`) hon. In variables ko access aur modify karne ke liye getter aur setter methods likhein.

```python
class Student:
    def __init__(self, name, roll_number):
        self.__name = name
        self.__roll_number = roll_number

    # Getter methods
    def get_name(self):
        return self.__name

    def get_roll_number(self):
        return self.__roll_number

    # Setter methods
    def set_name(self, name):
        self.__name = name

    def set_roll_number(self, roll_number):
        self.__roll_number = roll_number


student1 = Student("Ali", 101)
print("Name:", student1.get_name())
print("Roll number:", student1.get_roll_number())

student1.set_name("Ahmed")
student1.set_roll_number(102)
print("Updated name:", student1.get_name())
print("Updated roll number:", student1.get_roll_number())
```

**Q2:** Aik `BankAccount` class banayein jismein `__balance` private variable ho. Aik `deposit()` aur aik `withdraw()` method banayein, check karein ke withdraw karte waqt balance zero se kam na ho.

```python
class BankAccount:
    def __init__(self, balance=0):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposited:", amount)
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance.")

    def get_balance(self):
        return self.__balance


account = BankAccount(1000)
account.deposit(500)
account.withdraw(300)
account.withdraw(2000)
print("Current balance:", account.get_balance())
```

**Q3:** Aik class banayein `Employee` jismein salary private ho. Agar koi salary ko minus (negative) value set karne ki koshish kare, toh setter method 'Invalid Salary' ka error message show kare.

```python
class Employee:
    def __init__(self, salary):
        self.salary = salary  # Calls the setter below

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, amount):
        if amount < 0:
            raise ValueError("Invalid Salary")
        self.__salary = amount


employee1 = Employee(30000)
print("Salary:", employee1.salary)

try:
    employee1.salary = -5000
except ValueError as error:
    print(error)
```

**Q4:** Protected variable (`_`) aur Private variable (`__`) ke beech kya farq hota hai? Aik simple code ke zariye dono ka use-case dikhayein.

```python
class Example:
    def __init__(self):
        self._protected_value = "Protected variable"
        self.__private_value = "Private variable"

    def show_private_value(self):
        # A private variable can be accessed inside its own class.
        print(self.__private_value)


obj = Example()
print(obj._protected_value)  # Protected: accessible, but intended for class/subclass use
obj.show_private_value()     # Access private value through a class method

# obj.__private_value would cause an AttributeError because of name mangling.
```

**Q5:** Encapsulation ka real-world faida kya hai? Private variable ko class ke bahar se direct access karne par kaunsa error aata hai? Code likh kar check karein.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance

    def get_balance(self):
        return self.__balance


account = BankAccount(1000)
print("Balance through method:", account.get_balance())

try:
    print(account.__balance)  # Direct access is not allowed
except AttributeError as error:
    print("Error:", error)

# Benefit: encapsulation protects data and lets the class control how it is changed.
```

**Q6:** Single Inheritance ka use karte hue aik parent class `Vehicle` (with brand attribute) aur child class `Car` (with model attribute) banayein. Car ka object bana kar dono attributes print karein.

**Q7:** Multilevel Inheritance ka use karein. Class `Animal` -> Class `Mammal` -> Class `Dog`. Dog class ke object se parent classes ke features ko access karein.

```python
class Animal:
    def eat(self):
        print("Animal eats food.")


class Mammal(Animal):
    def feed_milk(self):
        print("Mammals feed milk to their babies.")


class Dog(Mammal):
    def bark(self):
        print("Dog barks.")


dog1 = Dog()
dog1.eat()         # Inherited from Animal
dog1.feed_milk()   # Inherited from Mammal
dog1.bark()        # Defined in Dog
```

**Q9:** Child class ke constructor (`__init__`) ke andar parent class ke constructor ko call karne ke liye `super()` ka use kaise hota hai? Aik practical example likhein.

```python
class Person:
    def __init__(self, name):
        self.name = name


class Student(Person):
    def __init__(self, name, roll_number):
        super().__init__(name)  # Calls Person's constructor
        self.roll_number = roll_number


student1 = Student("Sara", 12)
print("Name:", student1.name)
print("Roll number:", student1.roll_number)
```

**Q10:** Hierarchical Inheritance kya hoti hai? Aik parent class `Shape` banayein aur us se do alag child classes `Circle` aur `Square` banakar samjhayein.

```python
class Shape:
    def describe(self):
        print("This is a shape.")


class Circle(Shape):
    def draw(self):
        print("Drawing a circle.")


class Square(Shape):
    def draw(self):
        print("Drawing a square.")


circle1 = Circle()
square1 = Square()

circle1.describe()
circle1.draw()
square1.describe()
square1.draw()

# Hierarchical inheritance: Circle and Square both inherit from Shape.
```

**Q11:** Method Overriding kya hai? Parent class `Bird` banayein jisme `fly()` method ho jo print kare 'Most birds can fly'. Child class `Ostrich` banayein jo is method ko override kare aur print kare 'Ostriches cannot fly'.

```python
class Bird:
    def fly(self):
        print("Most birds can fly")


class Ostrich(Bird):
    def fly(self):  # Overrides Bird.fly()
        print("Ostriches cannot fly")


bird1 = Bird()
ostrich1 = Ostrich()
bird1.fly()
ostrich1.fly()
```

**Q12:** Polymorphism function ke zariye kaise achieve hoti hai? Aik function banayein `make_sound(animal_object)` jo har tarah ke animal class ka sound method automatically call kare.

```python
class Dog:
    def sound(self):
        print("Dog says: Woof!")


class Cat:
    def sound(self):
        print("Cat says: Meow!")


class Cow:
    def sound(self):
        print("Cow says: Moo!")


def make_sound(animal_object):
    # Python calls the sound() method of the object passed in.
    animal_object.sound()


make_sound(Dog())
make_sound(Cat())
make_sound(Cow())
```

**Q13:** Operator Overloading kya hoti hai? Do objects ko `+` operator ke zariye add karne ke liye `__add__` magic method ka use kaise karte hain? Code likhein.

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)


num1 = Number(10)
num2 = Number(20)
result = num1 + num2  # Uses the __add__() method
print("Sum:", result.value)
```

**Q15:** Kya Python default roop se Method Overloading (same method name with different arguments) support karta hai? Agar nahi, toh hum same name ke method ko multiple arguments ke sath kaise handle karte hain? Code likhein.

```python
# Python does not support traditional method overloading by defining
# multiple methods with the same name. A common solution is a default argument.

class Calculator:
    def add(self, a, b, c=0):
        return a + b + c


calculator = Calculator()
print(calculator.add(2, 3))       # Two arguments
print(calculator.add(2, 3, 4))    # Three arguments
```

**Q16:** `abc` module ka use karte hue aik abstract class `Computer` banayein jisme abstract method `process()` ho. Is class ka object banane ki koshish karein aur dekhein kya hota hai.

```python
from abc import ABC, abstractmethod


class Computer(ABC):
    @abstractmethod
    def process(self):
        pass


# An abstract class cannot be instantiated directly.
try:
    computer1 = Computer()
except TypeError as error:
    print("Error:", error)
```

**Q18:** Aik abstract class `Car` banayein jisme aik concrete (normal) method `start_engine()` ho aur aik abstract method `drive()` ho. Child class bana kar dono ko test karein.

```python
from abc import ABC, abstractmethod


class Car(ABC):
    def start_engine(self):
        print("Engine started.")  # Concrete (normal) method

    @abstractmethod
    def drive(self):
        pass


class SportsCar(Car):
    def drive(self):
        print("The car is moving.")


car1 = SportsCar()
car1.start_engine()
car1.drive()
```

**Q20:** Aik real-world scenario (jaise Payment System) design karein jahan `Payment` aik abstract class ho aur `CreditCardPayment` aur `UPIPayment` uski subclasses hon.

```python
from abc import ABC, abstractmethod


class Payment(ABC):
    @abstractmethod
    def pay(self, amount):
        pass


class CreditCardPayment(Payment):
    def pay(self, amount):
        print("Paid", amount, "using Credit Card.")


class UPIPayment(Payment):
    def pay(self, amount):
        print("Paid", amount, "using UPI.")


def checkout(payment_method, amount):
    payment_method.pay(amount)


credit_card = CreditCardPayment()
upi = UPIPayment()

checkout(credit_card, 1500)
checkout(upi, 750)
```
